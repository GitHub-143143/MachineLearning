```python
import pandas as pd
import numpy as np
from sklearn.linear_model import LinearRegression
import matplotlib.pyplot as plt
```

```python
data = pd.read_csv("linearprices.csv")
data.head()
```

```python
plt.xlabel("area")
plt.ylabel("price")
plt.scatter(data.area,data.price,color="green")
```

```python
reg = LinearRegression()
reg.fit(data[['area']],data.price)
```

```python
reg.predict([[3300]])
```

```python
reg.coef_
```

```python
reg.intercept_
```

```python
d = pd.read_csv("areas.csv")
```

```python
reg.predict(d)
```

```python
p = reg.predict(d)
d['prices'] = p
d.head()
```

```python
d.to_csv("prediction.csv",index='False')
```
